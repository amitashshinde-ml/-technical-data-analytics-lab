```python

from math import radians, sin, cos, asin, sqrt
my_lat = 28.6517178
my_lon = 77.2219388

iss_lat = '29.6799'
iss_long = '80.4634'

my_lat, my_lon, iss_lat, iss_long = map(
    radians,
    [float(my_lat), float(my_lon), float(iss_lat), float(iss_long)]
)

dlat = iss_lat - my_lat
dlon = iss_long - my_lon

a = sin(dlat/2)**2 + cos(my_lat) * cos(iss_lat) * sin(dlon/2)**2

distance = 6371 * 2 * asin(sqrt(a))
print(distance)
```
