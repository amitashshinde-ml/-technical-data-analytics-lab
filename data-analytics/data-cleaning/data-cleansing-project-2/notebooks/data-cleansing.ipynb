```python
# Column Name Cleanup
df.columns=df.columns.str.lower().replace(r"[\(\)\£]","",regex=True).str.strip().str.replace(" ","_")
df.columns
```

```python
# Cleanup spaces in data
for i in df.select_dtypes(include=["str"]):
    df[i]= df[i].str.strip()
```

```python
df.isna().sum()
```

```python
# Separated null 'primary key data' for stakeholder discussion
df_valid= df.loc[df["order_id"].isna()==False].copy()
df_quarentine=df.loc[df["order_id"].isna()].copy()
print(df_valid.shape)
print(df_quarentine.shape)
```

```python
df.head()
```

```python
# Dropped NULL quantity, unit price and customer name records as no replace logic possible
df_valid.dropna(subset=["quantity","unit_price","customer_name","order_date"],inplace=True)
df_valid.isna().sum()
```

```python
df_valid["discount_rate"].unique()
df_valid["discount_rate"]=df_valid["discount_rate"].fillna("0.0")
```

```python
df_valid.isna().sum()
```

```python
df_valid["payment_type"].unique()
```

```python
df_valid["region"]=df["region"].fillna("N/A")
df_valid["sales_channel"]=df["sales_channel"].fillna("unknown")  # as this one of the valid channel present on other records
df_valid["product_category"]=df["product_category"].fillna("N/A")  # as this one of the valid category present on other records
df_valid["order_status"]=df["order_status"].fillna("N/A")
df_valid["payment_type"]=df["payment_type"].fillna("N/A")
df_valid["email_address"]=df["email_address"].fillna("N/A")
df_valid["order_date"]=df["order_date"].dropna()
```

```python
for i in df_valid.select_dtypes(include=["str"]):
    if i != "email_address":
        print(i)
        df_valid[i]=df_valid[i].str.replace(r"[\?\*]",'',regex=True)
```

```python
df_valid.loc[df_valid["customer_name"]=='',"customer_name"]='Not available'
df_valid.loc[df_valid["quantity"]=='3 units',"quantity"]='3'
df_valid["discount_rate"]= df_valid["discount_rate"].str.replace("10%","1")

for i in df_valid.select_dtypes(include=["str"]):
    if i != "email_address":
        print(i)
        df_valid[i] = df_valid[i].str.replace(r"[/]", '', regex=True)
```

```python
#Finding unique values in each column to find if there are any special characters in String columns
for i in df_valid.select_dtypes(include=["str"]):
    if i != "email_address":
        print(f"{i} unique value")
        print(df_valid[i])
```

```python
for i in df_valid.select_dtypes(include=["str"]):
    if i != "email_address":
        print(i)
        df_valid[i]=df_valid[i].str.replace(r"[/]",'',regex=True)
```

```python
df_valid.info()
```

```python
# Change data type to numeric
df_valid=df_valid.astype({"quantity":"int","unit_price":"float","discount_rate":"float"})
df_valid.info()
df_valid.isna().sum()
```

```python
# Cleanup age
df_valid["age"]=df_valid["age"].fillna('unknown')
df_valid["age"]=df_valid["age"].str.strip()
df_valid.loc[df_valid["age"]=='','age']='unknown'
df_valid["age"].unique()
```

```python
# cleanup unit_price

df_valid["unit_price"].unique()
df_valid["unit_price"]=df_valid["unit_price"].str.strip()
df_valid["unit_price"]= df_valid["unit_price"].str.replace("£",'',regex=False)
df_valid= df_valid.loc[df_valid["unit_price"]!='unknown']
```

```python
# Clean up of order date
df_valid["order_date"].sort_values().unique()
df_valid.loc[df_valid["order_date"]=='31122024','order_date']='2024-12-31' # Fixed incorrectly formatted date
df_valid= df_valid.loc[df_valid["order_date"]!='2024-02-30']  # deleted in records with valid dates
df_valid=df_valid[df_valid["order_date"]!='unknown']
df_valid["order_date"]=pd.to_datetime(df_valid["order_date"])
```
