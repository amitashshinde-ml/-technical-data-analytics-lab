```python
# Records with columns country us null
print(hotel_booking_df[hotel_booking_df["country"].isna()].count())
hotel_booking_df[hotel_booking_df["country"].isna()]
```

```python
hotel_booking_df["country"] = hotel_booking_df["country"].fillna("Unknown")
```

```python
hotel_booking_df["company"].isna().sum()/len(hotel_booking_df)*100
```

```python
# Deleting Company column as it has 94 % null values and not providing any value
hotel_booking_df.drop(columns={"company"},inplace=True)
hotel_booking_df.columns
```

```python
# Verifying if all nulls are addressed
hotel_booking_df.isna().sum()
```

```python
hotel_booking_df.dtypes
```

```python
# Changing to right data types
hotel_booking_df=hotel_booking_df.astype({"is_canceled":"boolean", "no_of_children":"int","agent":"int","is_repeated_guest":"boolean"})
hotel_booking_df.dtypes
```

```python
# Bin the values in specifc groups
hotel_booking_df["lead_time"].unique()
hotel_booking_df["lead_time"].describe()

bins=[0,100,200,300,400,500,600,700,800]
labels =["0-100","101-200","201-300","301-400","401-500","501-600","601-700","701-800"]

hotel_booking_df["lead_time_bins"]=pd.cut(hotel_booking_df["lead_time"],bins=bins,labels=labels)

print(hotel_booking_df[["lead_time","lead_time_bins"]].sort_values(["lead_time"]))
```

```python
# Fixed nulls on new Bins column
hotel_booking_df["lead_time_bins"]=hotel_booking_df["lead_time_bins"].fillna("0-100")
print(hotel_booking_df[["lead_time","lead_time_bins"]].sort_values(["lead_time"]))
```

```python
# Move 'Lead time bin' column near 'Lead time' column
hotel_booking_df.dtypes

move_lt_column = hotel_booking_df.pop("lead_time_bins")
hotel_booking_df.insert(3,"lead_time_bins", move_lt_column)
hotel_booking_df.head()
```

```python
#Changed string data type of Date column to datetime
hotel_booking_df["reservation_status_date"]=pd.to_datetime(hotel_booking_df["reservation_status_date"],yearfirst=True)
hotel_booking_df["reservation_status_date"].dtypes
hotel_booking_df["reservation_status_date"].head(5)
```
